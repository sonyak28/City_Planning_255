```python
from IPython.display import HTML, display
display(HTML("""
<style>
  .sr-only{position:absolute;width:1px;height:1px;padding:0;overflow:hidden;clip:rect(0,0,0,0);white-space:nowrap;border:0}
  .eq-tab-bar{display:flex;gap:4px;border-bottom:1px solid #e0ddd6;margin-bottom:1.5rem;flex-wrap:wrap}
  .eq-tab-btn{background:none;border:none;border-bottom:2px solid transparent;padding:8px 18px;font-size:14px;color:#666;cursor:pointer;margin-bottom:-1px;font-family:inherit}
  .eq-tab-btn.active{color:#1a1a1a;border-bottom-color:#3266ad;font-weight:500}
  .eq-tab-panel{display:none}
  .eq-tab-panel.active{display:block}
  .eq-metric-grid{display:grid;grid-template-columns:repeat(auto-fit,minmax(140px,1fr));gap:10px;margin-bottom:1.5rem}
  .eq-metric-card{background:#f7f6f1;border-radius:8px;padding:12px 14px}
  .eq-metric-label{font-size:12px;color:#888;margin:0 0 4px;font-family:inherit}
  .eq-metric-val{font-size:22px;font-weight:500;color:#1a1a1a;margin:0;font-family:inherit}
  .eq-metric-sub{font-size:11px;color:#aaa;margin:2px 0 0;font-family:inherit}
  .eq-section{font-size:15px;font-weight:500;color:#1a1a1a;margin:1.5rem 0 .75rem;border-bottom:1px solid #e0ddd6;padding-bottom:6px;font-family:inherit}
  .eq-tbl{width:100%;border-collapse:collapse;font-size:13px;margin-bottom:1.5rem;font-family:inherit}
  .eq-tbl th{text-align:left;padding:7px 10px;color:#666;font-weight:500;border-bottom:1px solid #e0ddd6;background:#f7f6f1;font-family:inherit}
  .eq-tbl td{padding:7px 10px;border-bottom:1px solid #e0ddd6;color:#1a1a1a;font-family:inherit}
  .eq-tbl tr:last-child td{border-bottom:none}
  .eq-badge{display:inline-block;font-size:11px;padding:2px 7px;border-radius:4px;font-weight:500;font-family:inherit}
  .eq-sig-high{background:#E6F1FB;color:#0C447C}
  .eq-sig-med{background:#EAF3DE;color:#27500A}
  .eq-sig-low{background:#FAEEDA;color:#633806}
  .eq-sig-ns{background:#f0ede6;color:#888}
  .eq-bar-wrap{margin-bottom:1.1rem}
  .eq-bar-label{display:flex;justify-content:space-between;font-size:13px;margin-bottom:4px;font-family:inherit}
  .eq-bar-track{background:#f0ede6;border-radius:4px;height:10px;overflow:hidden}
  .eq-bar-fill{height:100%;border-radius:4px}
  .eq-corr-row{display:flex;align-items:center;gap:10px;margin-bottom:14px;font-size:13px;font-family:inherit}
  .eq-note{font-size:12px;color:#999;margin-top:-0.75rem;margin-bottom:1rem;font-family:inherit}
  .eq-callout{font-size:13px;color:#666;line-height:1.7;background:#f7f6f1;border-radius:8px;padding:14px 16px;font-family:inherit}
  .eq-explainer{font-size:13px;color:#444;line-height:1.75;background:#f0f5fb;border-left:3px solid #3266ad;border-radius:0 8px 8px 0;padding:14px 16px;margin-bottom:1.2rem;font-family:inherit}
  .eq-takeaway{font-size:13px;color:#555;line-height:1.75;background:#fdf6ec;border-left:3px solid #BA7517;border-radius:0 8px 8px 0;padding:14px 16px;margin-top:1rem;font-family:inherit}
</style>

<div style='font-family: -apple-system, BlinkMacSystemFont, Segoe UI, sans-serif;'>

<div class='eq-tab-bar'>
  <button class='eq-tab-btn active' onclick='eqShowTab("perm",this)'>Permutation tests</button>
  <button class='eq-tab-btn' onclick='eqShowTab("demo",this)'>Demographics</button>
  <button class='eq-tab-btn' onclick='eqShowTab("unmet",this)'>Unmet need</button>
  <button class='eq-tab-btn' onclick='eqShowTab("div",this)'>Amenity diversity</button>
</div>

<!-- ==================== PERMUTATION TESTS TAB ==================== -->
<div id='eq-tab-perm' class='eq-tab-panel active'>

  <div class='eq-explainer'>
    <strong>What are permutation tests, and why use them?</strong><br>
    A permutation test asks a simple question: <em>could the difference we see between two groups have happened just by chance?</em>
    We randomly shuffle the station labels 10,000 times and check how often a gap as large as the one we observed shows up by accident.
    If that almost never happens (p &lt; 0.05), we treat the difference as meaningful rather than random noise.
    This method is especially useful here because it does not rely on strong assumptions about the data, which makes it a good fit for a small real-world dataset like ours.<br><br>
    <strong>How “core” and “peripheral” were actually defined:</strong>
    These labels were based on FY2025 ridership across BART and Caltrain stations rather than geography alone. In this project, <strong>core</strong> stations are those consistently 
    identified as higher-ridership stations, while <strong>peripheral</strong> stations are those in the lower-ridership group.
  </div>

  <div class='eq-metric-grid'>
    <div class='eq-metric-card'><p class='eq-metric-label'>Core mean amenities</p><p class='eq-metric-val'>21.9</p><p class='eq-metric-sub'>n = 40 stations</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Peripheral mean</p><p class='eq-metric-val'>9.8</p><p class='eq-metric-sub'>n = 39 stations</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Raw gap</p><p class='eq-metric-val'>12.0</p><p class='eq-metric-sub'>Core minus peripheral</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Glass Delta (total)</p><p class='eq-metric-val'>2.17</p><p class='eq-metric-sub'>Very large effect</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Significant after FDR</p><p class='eq-metric-val'>6 / 10</p><p class='eq-metric-sub'>B-H correction</p></div>
  </div>

  <div class='eq-callout' style='margin-bottom:1.5rem'>
    <strong style='color:#1a1a1a'>Bottom line:</strong> Core stations have on average <strong>more than twice as many</strong> nearby amenities as peripheral stations (21.9 vs. 9.8).
    This gap held up across 6 out of 10 amenity types after correcting for multiple comparisons.
    The effect size — Glass’s Delta of 2.17 — is extremely large: the average core station is better served than roughly 98% of peripheral stations.
    This is not a small difference. It points to a real structural gap.
  </div>

  <p class='eq-section'>Core vs peripheral: permutation test results (10,000 resamples)</p>
  <p style='font-size:12px;color:#999;margin-bottom:10px;font-family:inherit'>
    <strong>Reading the columns:</strong> “p (raw)” = probability the observed gap happened by chance.
    “q (FDR)” = a stricter threshold after correcting for running 10 tests at once using the Benjamini-Hochberg method.
    “Glass Delta” = the practical size of the difference (0.5+ = moderate, 0.8+ = large, 2.0+ = very large).
    “n.s.” = not statistically significant after correction.
  </p>
  <table class='eq-tbl' id='eq-perm-tbl'>
    <thead><tr>
      <th>Amenity category</th><th>Core mean</th><th>Peripheral mean</th>
      <th>Difference</th><th>p (raw)</th><th>q (FDR)</th><th>Glass Delta</th><th>Result</th>
    </tr></thead>
    <tbody></tbody>
  </table>

  <p class='eq-section'>Glass Delta by amenity category</p>
  <p style='font-size:12px;color:#999;margin-bottom:10px;font-family:inherit'>Delta &ge; 0.8 = large effect; &ge; 2.0 = very large. Gray bars = not statistically significant after FDR correction.</p>
  <div id='eq-effect-bars'></div>

  <div class='eq-takeaway'>
    <strong>Real-world meaning:</strong>
    The biggest gaps are in <strong>convenience stores</strong> (D = 2.47) and <strong>clinics</strong> (D = 2.00),
    which are the kinds of everyday resources riders often depend on. A peripheral station like Pittsburg/Bay Point or Antioch may have a BART stop,
    but the surrounding area offers far fewer places to pick up groceries, fill a prescription, or see a doctor without a car.
    Hospitals and childcare did not show significant gaps, likely because both are relatively scarce across the network.
    In other words, equality there may reflect low access everywhere, not good access anywhere.
  </div>
</div>

<!-- ==================== DEMOGRAPHICS TAB ==================== -->
<div id='eq-tab-demo' class='eq-tab-panel'>

  <div class='eq-explainer'>
    <strong>Are stations serving higher proportions of non-white residents, low-income households, or zero-vehicle households more likely to be amenity-poor?</strong><br>
    Having a transit station is not the same thing as having equitable access.
    If the stations serving the most disadvantaged communities also have the fewest nearby services,
    that creates a compounding disadvantage — not just a transit gap, but a gap in everyday livability.<br><br>
    We used <strong>Spearman rank correlations</strong> to test whether three demographic variables are related to how amenity-rich a station’s neighborhood is.
    Spearman is more appropriate than Pearson here because our amenity counts are skewed and not normally distributed.
    It works off rankings rather than raw values, so it is less sensitive to outliers.
    A positive rho means more of that group is associated with more amenities; a negative rho means more of that group is associated with fewer amenities.
  </div>

  <div class='eq-metric-grid'>
    <div class='eq-metric-card'><p class='eq-metric-label'>Core: no vehicle</p><p class='eq-metric-val'>22.5%</p><p class='eq-metric-sub'>Transit-dependent</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Peripheral: no vehicle</p><p class='eq-metric-val'>6.5%</p><p class='eq-metric-sub'>Transit-dependent</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Core median income</p><p class='eq-metric-val'>$119k</p><p class='eq-metric-sub'>Household</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Peripheral median income</p><p class='eq-metric-val'>$164k</p><p class='eq-metric-sub'>Household</p></div>
  </div>

  <p class='eq-section'>Spearman correlations with total amenity count</p>
  <p style='font-size:12px;color:#999;margin-bottom:10px;font-family:inherit'>
    Bars extending <strong>right of center</strong> mean more of that group is associated with more amenities.
    Bars extending <strong>left of center</strong> mean more of that group is associated with fewer amenities.
    Significance badges reflect q-values after FDR correction.
  </p>
  <div id='eq-corr-bars' style='margin-bottom:1.5rem'></div>

  <p class='eq-section'>What the correlations mean in plain language</p>
  <div class='eq-callout' style='margin-bottom:1rem'>
    <p style='margin:0 0 10px'>
      <strong style='color:#1a1a1a'>% zero-vehicle households (rho = +0.476, p &lt; 0.001) — strongest signal:</strong>
      Stations with more car-free households nearby tend to have more amenities around them.
      On the surface, that sounds encouraging, but it is mostly a <em>density effect</em>:
      car-free households cluster in the urban core, where things are already closer together.
      The Unmet Need tab shows that some of these same stations still rank poorly for actual access gaps,
      which means physical proximity alone does not guarantee that the available services are affordable, safe, or sufficient.
    </p>
    <p style='margin:0 0 10px'>
      <strong style='color:#1a1a1a'>% non-white residents (rho = &minus;0.243, p = 0.047) — equity concern:</strong>
      There is a statistically significant negative relationship here: stations in majority non-white neighborhoods tend to have fewer services within walking distance.
      In practical terms, a 10-percentage-point increase in non-white share is associated with roughly 2–3 fewer walkable amenities.
      That can mean the difference between being able to reach a pharmacy or grocery store on foot and needing a car or another transit trip.
    </p>
    <p style='margin:0 0 0'>
      <strong style='color:#1a1a1a'>Median household income (rho = &minus;0.064, p = 0.575) — not significant:</strong>
      Income is <em>not</em> a meaningful predictor in this dataset.
      That is important because it suggests the amenity gap is not simply a wealthy-versus-poor neighborhood story.
      In the Bay Area, the pattern seems to be shaped more by urban form and racial geography than by income alone.
      Interventions based only on income thresholds could miss some of the places with the greatest need.
    </p>
  </div>

  <div class='eq-takeaway'>
    <strong>Answer: Partially yes, and the pattern matters.</strong>
    Stations in majority non-white neighborhoods face a measurable amenity disadvantage.
    Zero-vehicle households show the opposite pattern, but that mostly reflects urban density rather than true equity.
    Income alone does not explain the gap, which suggests that equity-focused planning needs to account for racial geography and long-term neighborhood investment patterns, not just income.
  </div>
</div>

<!-- ==================== UNMET NEED TAB ==================== -->
<div id='eq-tab-unmet' class='eq-tab-panel'>

  <div class='eq-explainer'>
    <strong>What is the “unmet need index” and why does it matter?</strong><br>
    A station can have a handful of amenities nearby and still not serve its riders well if the surrounding population is highly transit-dependent and those amenities are limited relative to that level of need.
    The <strong>unmet need index</strong> combines two things into one score:
    how many people around a station are car-free (the need) and how few amenities that station has nearby (the gap).
    A high score means many people who depend on this station have relatively little within walking distance.<br><br>
    The <strong>Gini coefficient</strong> is borrowed from economics, where it measures inequality
    (0 = perfectly equal, 1 = everything concentrated in one place).
    Applied here, a Gini of 0 would mean every station has the same amenity count,
    while a Gini of 1 would mean amenities are concentrated at a single station.
    Our network Gini of <strong>0.428</strong> suggests that access to amenities is unevenly distributed across the Bay Area transit network.
  </div>

  <div class='eq-metric-grid'>
    <div class='eq-metric-card'><p class='eq-metric-label'>Gini: full network</p><p class='eq-metric-val'>0.428</p><p class='eq-metric-sub'>95% CI: 0.361&ndash;0.478</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Gini: core</p><p class='eq-metric-val'>0.410</p><p class='eq-metric-sub'>95% CI: 0.338&ndash;0.461</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Gini: peripheral</p><p class='eq-metric-val'>0.315</p><p class='eq-metric-sub'>95% CI: 0.238&ndash;0.385</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Highest unmet need</p><p class='eq-metric-val'>Coliseum</p><p class='eq-metric-sub'>Index score: 0.801</p></div>
  </div>

  <div class='eq-callout' style='margin-bottom:1.5rem'>
    <strong style='color:#1a1a1a'>Why the peripheral Gini is lower — and why that is not necessarily good news:</strong>
    Peripheral stations have a lower Gini (0.315 vs. 0.410 for core) because they are <em>uniformly</em> underserved.
    There is less inequality among them because most of them are similarly sparse.
    Core stations show more internal variation: some are very well-served (such as Montgomery St. or 12th St. Oakland),
    while others are surprisingly under-resourced (such as Coliseum or Bay Fair).
    The Gini measures spread, not overall level. Peripheral stations may look more equal simply because access is low across the board.
  </div>

  <p class='eq-section'>Top 10 stations by unmet need index</p>
  <p style='font-size:12px;color:#999;margin-bottom:10px;font-family:inherit'>
    Ranked by how much the car-free population around a station exceeds what the surrounding neighborhood can offer on foot.
    A score near 1.0 means a very large gap between need and supply. Higher scores point to the strongest equity concerns.
  </p>
  <table class='eq-tbl' id='eq-unmet-tbl'>
    <thead><tr>
      <th>#</th><th>Station</th><th>Agency</th><th>Type</th>
      <th>Amenities</th><th>% No vehicle</th><th>Unmet need index</th>
    </tr></thead>
    <tbody></tbody>
  </table>
  <p class='eq-note'>Unmet need index = % no-vehicle &times; (1 &minus; normalized amenity count), min-max normalized across all stations.</p>

  <p class='eq-section'>Gini coefficient reference</p>
  <div class='eq-callout'>
    A Gini of 0 means perfectly equal access; 1 means all amenities are concentrated at one station. The network-wide Gini of
    <strong style='color:#1a1a1a'>0.428</strong> indicates a fairly uneven distribution of amenities.
    The lower peripheral Gini (0.315) should not be read as better access. It mainly reflects that peripheral stations are consistently low in amenity access.
  </div>

  <div class='eq-takeaway'>
    <strong>Real-world meaning:</strong>
    Coliseum station in East Oakland tops the list with an unmet need score of 0.801.
    Nearly 30% of surrounding households are car-free, yet the station has only 3 nearby amenities.
    Bay Fair, MacArthur, and El Cerrito del Norte show a similar pattern: high transit dependence and low walkable supply.
    These are the places where targeted investments in grocery access, clinics, and pharmacies would likely have the strongest equity payoff.
  </div>
</div>

<!-- ==================== AMENITY DIVERSITY TAB ==================== -->
<div id='eq-tab-div' class='eq-tab-panel'>

  <div class='eq-explainer'>
    <strong>Why diversity of amenities matters, not just raw count</strong><br>
    A station surrounded by 10 convenience stores is not equivalent to one with a clinic, a pharmacy, a grocery store, and a park,
    even if the total number of amenities is the same. <strong>Shannon entropy</strong> measures how evenly amenities are spread across different categories.
    A high entropy score means a strong mix of types; a low score means the neighborhood is dominated by only one or two kinds of amenities.<br><br>
    We used a <strong>Mann-Whitney U test</strong> to compare entropy between core and peripheral stations.
    This is a non-parametric alternative to the t-test and is appropriate here because entropy values are not normally distributed.
    The result (p = 0.041) suggests that the diversity gap between core and peripheral stations is unlikely to be due to chance.
  </div>

  <div class='eq-metric-grid'>
    <div class='eq-metric-card'><p class='eq-metric-label'>Core mean entropy</p><p class='eq-metric-val'>1.335</p><p class='eq-metric-sub'>Shannon entropy</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Peripheral mean entropy</p><p class='eq-metric-val'>1.124</p><p class='eq-metric-sub'>Shannon entropy</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Difference</p><p class='eq-metric-val'>0.212</p><p class='eq-metric-sub'>Core minus peripheral</p></div>
    <div class='eq-metric-card'><p class='eq-metric-label'>Mann-Whitney p</p><p class='eq-metric-val'>0.041</p><p class='eq-metric-sub'>U = 989.0</p></div>
  </div>

  <div class='eq-callout' style='margin-bottom:1.5rem'>
    <strong style='color:#1a1a1a'>Bottom line:</strong>
    Core stations are not only surrounded by <em>more</em> amenities — they also tend to have a <em>more balanced mix</em> of them (p = 0.041).
    Peripheral riders do not just have fewer options; they also have less variety.
    That matters because if the only thing nearby is a convenience store,
    riders may still need to travel farther for a prescription, a doctor, or basic groceries.
  </div>

  <p class='eq-section'>Top 10 most diverse stations by amenity entropy</p>
  <p style='font-size:12px;color:#999;margin-bottom:10px;font-family:inherit'>
    <strong>How to read:</strong> Entropy is highest when amenities are spread evenly across all 8 categories (max possible &asymp; 2.08).
    Heat shading by column shows relative abundance — darker blue means more of that category relative to other stations.
    A station can rank highly here even without the highest total count; what matters is balance.
  </p>
  <table class='eq-tbl' id='eq-div-tbl'>
    <thead><tr>
      <th>Station</th><th>Total</th><th>Entropy</th>
      <th>Groc</th><th>Park</th><th>Clinic</th>
      <th>Phar</th><th>Care</th><th>Doc</th><th>Hosp</th><th>Conv</th>
    </tr></thead>
    <tbody></tbody>
  </table>
  <p class='eq-note'>Shannon entropy across amenity categories. Max theoretical entropy = ln(8) &approx; 2.08 (perfectly even spread across all 8 types).</p>

  <div class='eq-takeaway'>
    <strong>Real-world meaning:</strong>
    Fremont ranks highest on diversity despite having only 15 total amenities,
    because those 15 are spread across 7 different categories.
    A rider there can reach a pharmacy, a grocery store, a doctor, and a hospital.
    Compare that with a station that has 20 amenities but 15 of them are convenience stores:
    the raw count looks strong, but the range of daily needs met is much narrower.
    For transit equity, diversity of access matters just as much as volume.
  </div>
</div>

</div>

<script>
(function() {
  window.eqShowTab = function(id, btn) {
    document.querySelectorAll('.eq-tab-panel').forEach(function(p) { p.classList.remove('active'); });
    document.querySelectorAll('.eq-tab-btn').forEach(function(b) { b.classList.remove('active'); });
    document.getElementById('eq-tab-' + id).classList.add('active');
    btn.classList.add('active');
  };

  var permData = [
    {name:'Total amenities', core:21.9, peri:9.8,  diff:12.0, p:0.0002, q:0.0010, d:2.17, sig:true},
    {name:'Grocery stores',  core:2.5,  peri:1.5,  diff:1.1,  p:0.0142, q:0.0284, d:0.74, sig:true},
    {name:'Parks',           core:5.5,  peri:3.6,  diff:1.9,  p:0.0222, q:0.0370, d:0.57, sig:true},
    {name:'Clinics',         core:2.5,  peri:0.6,  diff:1.9,  p:0.0002, q:0.0010, d:2.00, sig:true},
    {name:'Pharmacies',      core:1.2,  peri:0.6,  diff:0.7,  p:0.0058, q:0.0145, d:1.02, sig:true},
    {name:'Convenience',     core:7.8,  peri:2.5,  diff:5.2,  p:0.0004, q:0.0013, d:2.47, sig:true},
    {name:'Hospital',        core:0.2,  peri:0.1,  diff:0.2,  p:0.1834, q:0.2182, d:0.65, sig:false},
    {name:'Childcare',       core:0.4,  peri:0.2,  diff:0.3,  p:0.0962, q:0.1374, d:0.64, sig:false},
    {name:'Doctors',         core:1.0,  peri:0.6,  diff:0.4,  p:0.1964, q:0.2182, d:0.34, sig:false},
    {name:'Kindergarten',    core:0.0,  peri:0.0,  diff:0.0,  p:1.0000, q:1.0000, d:null, sig:false}
  ];

  function sigBadge(r) {
    if (!r.sig) return '<span class="eq-badge eq-sig-ns">n.s.</span>';
    if (r.d >= 2.0) return '<span class="eq-badge eq-sig-high">D>2 ***</span>';
    if (r.d >= 0.8) return '<span class="eq-badge eq-sig-med">D>0.8 **</span>';
    return '<span class="eq-badge eq-sig-low">D>0.5 *</span>';
  }

  var ptbody = document.querySelector('#eq-perm-tbl tbody');
  permData.forEach(function(r) {
    var tr = document.createElement('tr');
    tr.innerHTML =
      '<td>' + r.name + '</td>' +
      '<td>' + r.core.toFixed(1) + '</td>' +
      '<td>' + r.peri.toFixed(1) + '</td>' +
      '<td>' + r.diff.toFixed(1) + '</td>' +
      '<td>' + r.p.toFixed(4) + '</td>' +
      '<td>' + r.q.toFixed(4) + '</td>' +
      '<td>' + (r.d !== null ? r.d.toFixed(2) : '-') + '</td>' +
      '<td>' + sigBadge(r) + '</td>';
    ptbody.appendChild(tr);
  });

  var barsEl = document.getElementById('eq-effect-bars');
  var sorted = permData.filter(function(r) { return r.d !== null; }).sort(function(a,b) { return b.d - a.d; });
  var maxD = 2.6;
  sorted.forEach(function(r) {
    var pct = Math.min((r.d / maxD) * 100, 100).toFixed(1);
    var col = r.sig ? (r.d >= 2.0 ? '#3266ad' : r.d >= 0.8 ? '#639922' : '#BA7517') : '#b4b2a9';
    var wrap = document.createElement('div');
    wrap.className = 'eq-bar-wrap';
    wrap.innerHTML =
      '<div class="eq-bar-label"><span style="color:#1a1a1a">' + r.name + '</span>' +
      '<span style="color:#888">D = ' + r.d.toFixed(2) + (r.sig ? '' : ' (n.s.)') + '</span></div>' +
      '<div class="eq-bar-track"><div class="eq-bar-fill" style="width:' + pct + '%;background:' + col + '"></div></div>';
    barsEl.appendChild(wrap);
  });

  var corrData = [
    {label:'% no vehicle',  rho:0.476,  q:'< 0.001', sig:true,  dir:1},
    {label:'% non-white',   rho:-0.243, q:'0.047',   sig:true,  dir:-1},
    {label:'Median income', rho:-0.064, q:'0.575',   sig:false, dir:-1}
  ];
  var corrEl = document.getElementById('eq-corr-bars');
  corrData.forEach(function(c) {
    var absPct = Math.abs(c.rho) * 100;
    var col = c.sig ? (c.dir > 0 ? '#3266ad' : '#D85A30') : '#b4b2a9';
    var barLeft = c.dir > 0 ? 50 : (50 - absPct);
    var sigSpan = c.sig ? '<span class="eq-badge eq-sig-high" style="font-size:10px">sig</span>' : '<span class="eq-badge eq-sig-ns" style="font-size:10px">n.s.</span>';
    var row = document.createElement('div');
    row.className = 'eq-corr-row';
    row.innerHTML =
      '<span style="min-width:110px;color:#1a1a1a">' + c.label + '</span>' +
      '<div style="flex:1;position:relative;height:8px;background:#f0ede6;border-radius:4px">' +
        '<div style="position:absolute;left:50%;top:0;width:1px;height:8px;background:#ccc"></div>' +
        '<div style="position:absolute;top:0;left:' + barLeft + '%;width:' + absPct + '%;height:8px;border-radius:4px;background:' + col + '"></div>' +
      '</div>' +
      '<span style="min-width:60px;font-weight:500;font-size:13px;color:#1a1a1a;text-align:right">rho = ' + c.rho.toFixed(3) + '</span>' +
      '<span style="min-width:110px;text-align:right;font-size:12px;color:#888">q = ' + c.q + ' ' + sigSpan + '</span>';
    corrEl.appendChild(row);
  });

  var unmetData = [
    {rank:1,  station:'Coliseum',             agency:'BART',     type:'core',       amen:3,  noveh:29.4, idx:0.801},
    {rank:2,  station:'Bay Fair',             agency:'BART',     type:'core',       amen:5,  noveh:16.0, idx:0.654},
    {rank:3,  station:'Pittsburg Center',     agency:'BART',     type:'core',       amen:3,  noveh:12.2, idx:0.577},
    {rank:4,  station:'El Cerrito del Norte', agency:'BART',     type:'core',       amen:8,  noveh:16.1, idx:0.538},
    {rank:5,  station:'Castro Valley',        agency:'BART',     type:'peripheral', amen:10, noveh:18.7, idx:0.488},
    {rank:6,  station:'MacArthur',            agency:'BART',     type:'core',       amen:10, noveh:18.2, idx:0.481},
    {rank:7,  station:'Balboa Park',          agency:'BART',     type:'core',       amen:10, noveh:17.3, idx:0.474},
    {rank:8,  station:'Lawrence',             agency:'Caltrain', type:'peripheral', amen:9,  noveh:14.7, idx:0.458},
    {rank:9,  station:'Rockridge',            agency:'BART',     type:'core',       amen:10, noveh:16.1, idx:0.452},
    {rank:10, station:'Sunnyvale',            agency:'Caltrain', type:'peripheral', amen:5,  noveh:9.1,  idx:0.447}
  ];
  var utbody = document.querySelector('#eq-unmet-tbl tbody');
  unmetData.forEach(function(r) {
    var typeBg  = r.type === 'core' ? '#E6F1FB' : '#FAEEDA';
    var typeCol = r.type === 'core' ? '#0C447C'  : '#633806';
    var idxCol  = r.idx >= 0.7 ? '#A32D2D' : r.idx >= 0.5 ? '#D85A30' : '#444';
    utbody.innerHTML +=
      '<tr>' +
      '<td style="color:#999">' + r.rank + '</td>' +
      '<td style="font-weight:500">' + r.station + '</td>' +
      '<td>' + r.agency + '</td>' +
      '<td><span class="eq-badge" style="background:' + typeBg + ';color:' + typeCol + '">' + r.type + '</span></td>' +
      '<td>' + r.amen + '</td>' +
      '<td>' + r.noveh + '%</td>' +
      '<td style="font-weight:500;color:' + idxCol + '">' + r.idx.toFixed(3) + '</td>' +
      '</tr>';
  });

  var divData = [
    {station:'Fremont',                      total:15, ent:1.845, groc:2, park:1,  clin:1, phar:3, care:0, doc:2, hosp:3, conv:1},
    {station:'Lake Merritt',                 total:38, ent:1.832, groc:6, park:5,  clin:6, phar:3, care:3, doc:3, hosp:0, conv:11},
    {station:'San Mateo',                    total:20, ent:1.817, groc:4, park:4,  clin:4, phar:1, care:0, doc:4, hosp:1, conv:2},
    {station:'Redwood City',                 total:21, ent:1.791, groc:4, park:5,  clin:3, phar:2, care:0, doc:1, hosp:1, conv:5},
    {station:'SF Caltrain Station',          total:25, ent:1.731, groc:3, park:9,  clin:5, phar:1, care:2, doc:2, hosp:0, conv:3},
    {station:'12th St. Oakland City Center', total:53, ent:1.728, groc:5, park:10, clin:7, phar:5, care:2, doc:4, hosp:0, conv:19},
    {station:'College Park',                 total:12, ent:1.720, groc:2, park:2,  clin:3, phar:0, care:1, doc:2, hosp:0, conv:1},
    {station:'Gilroy',                       total:16, ent:1.675, groc:2, park:2,  clin:0, phar:2, care:1, doc:3, hosp:0, conv:5},
    {station:'Montgomery St.',               total:49, ent:1.673, groc:4, park:14, clin:8, phar:5, care:1, doc:2, hosp:0, conv:14},
    {station:'Palo Alto',                    total:27, ent:1.625, groc:3, park:9,  clin:5, phar:3, care:0, doc:5, hosp:0, conv:1}
  ];
  function heatBg(val, max) {
    if (val === 0) return 'transparent';
    var p = val / max;
    if (p > 0.6) return '#B5D4F4';
    if (p > 0.25) return '#E6F1FB';
    return '#EAF3DE';
  }
  var divbody = document.querySelector('#eq-div-tbl tbody');
  divData.forEach(function(r) {
    divbody.innerHTML +=
      '<tr>' +
      '<td style="font-weight:500">' + r.station + '</td>' +
      '<td>' + r.total + '</td>' +
      '<td style="font-weight:500">' + r.ent.toFixed(3) + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.groc,6)  + '">' + r.groc + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.park,14) + '">' + r.park + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.clin,8)  + '">' + r.clin + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.phar,5)  + '">' + r.phar + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.care,3)  + '">' + r.care + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.doc,5)   + '">' + r.doc  + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.hosp,3)  + '">' + r.hosp + '</td>' +
      '<td style="text-align:center;background:' + heatBg(r.conv,19) + '">' + r.conv + '</td>' +
      '</tr>';
  });
})();
</script>
"""))
```
